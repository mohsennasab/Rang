```python
#@title Set up this notebook
import base64
import json
import subprocess
import sys
import types
import zipfile
import zlib
from pathlib import Path

IN_COLAB = False
try:
    from google.colab import files
    IN_COLAB = True
except ImportError:
    pass

try:
    import matplotlib
    import numpy
    import PIL
    import sklearn
except ImportError:
    subprocess.run([
        sys.executable, "-m", "pip", "install", "-q",
        "numpy", "pillow", "scikit-learn", "matplotlib"
    ], check=True)

PACKED_MODULE_SOURCES = "c-rNC4SU-*lJH;Qx_4)?p%_X2R;unLj@x+GP4gt#k6T%%QX*xsrbw2gVyiy?_jf)35Fja8?)L8e+CCdoBrq5ZW(G5Z0r0RlOXD<;qv_tj+PkLzJ}vw_3@rL!E&Xy~&EcV3gw{ttxv|PLjSKHOxlYd0FJWF-#mDd8z+3D5^5W|C8_Q1u>*{;~^%MLYrAgrmU|%BqX$9dtnnhtUTU$l6T*ZC~Z?6-7UWU2#gFjpNL28Ai<;R|7_q*LQ$ED_>fwvd%yDPv2(R?1}04H>*RV<VFaafk&brOeP!q{4bafUcfS5aJA&~}+-RyxO`OV9eS%0nwzEwi<26)DxfTcmMFeFQY3#L8Dmah=>n<sw}HMxy0{&<~*}S(pUa6C?=JFNu~}nwM4ty&+nuH$NBHP4OcG!UEaV?>tnGt2~A;ULO9k3XAf(szaz+eqkjUzWsRd;m66xAAkI84JA151N^lolX(<}lZgW`iZuQb+K!jOpeCjGbDey6eRc8q^Tp)t<wvMpL3(Lj?+F?f*qQU!=kMA}&0wHHEOMQEWqrugpTk)RgvrA&iI7be&^8b=AnsX4(1+C$(2VlCsL;o91q94ttYav+$`RO?D7-@u6>}7x^%tlVmDqV4geCMN0&)^#Bo<&YyQT4)Lqo*^*moY?!05uTod0<B<Ht8wm+#+BKEL~T@#)=<S8t&+r(Ty)@BEYhWBw<u_ooQ*hsodn;~&sMl6iif`)hkN>UxL0{^8N_p=))$gQJt9lj9yf_D;Kp2Pa2k_d2l}|ASinUjO){e}+X*dPlvZejgvZC;i^ZQFmjlZtv{i;DBoNdS|_}eit9lj=Bfk!`^u8Tql2fclr6kK#{&5p14-e>(W1`r>6&46#oN(1xdP}l<6eDnc53yz%4HT#KMb<GLJI*dhe_2J%>N!VY$i^4SbZ8_F@zbBI||KA43Dj4)H}~fsEPEB6RG~fe-UI^-HZ;Xb+o()@^u{4d<?P<62YhL7oB&+3+xn0Szbxy#;_rM~>t9MY+yG8=5yUn*+8F`Jmgse|)&%KU4a{3~hATM;xdF;r%QPP}#)U!XIJ_0v{Ir&%l30bdv!09rUd!aO-Pi4X7I40E>I?FFsqVA_R2^Ts%jtvG-o?7fV00P-I!JBE-KfKz~@McdYpeWP_TtqQdF`<7B9})?pl{cl4@%bnJrWie?Mc*B}f9$^rvAM}-AKQ^FukR{-<l>rbF$Gf<1xT@H;Dh{UH=KKEzgcM&ud#^Dll1maTljh~}Av==2s>4WyO2`3l^P;!n1gPEjd%V(s${Mxm?8H%pJ0}yafv2gCAUihmYufmHwPjg@>F1cJ4r8Nyf#1tv#dAbB(FzEn{it7|t;DFm-n8m(e|8(x<QFKJ5FW4Yf5omp5^$#8Ezbre)XXI0OzQN~He4H9=#FaZ%-}?G%9CM%#zxsA4-t~Z-&@FmfaaJMNI*yXi&+VDyqnVJ&g2iSg<pC5fbQCp~DP4C9nE&h<5);J!uzPp}Q!Nzk^}REYC>uV#z%Rh5_?904t-g1t$rxjS_#j>lY)Guyup&9@Iopxy9`p{nr!J#peQnKL#seP_6}yL>1AKst0@JT)RO>1r_NTfExL^MY=+y{`->rYQNCSH~O_8GG3UlraV~4fBCG<{DQFySM)4as@(5(~i<n#p4z)~-O2atBa3H6QwpU-aX5qyHO^O5Iy(6=%4qj%h@n&E#(FyTR;Md2<MRu!F1mHI}do>r+pMxEK|>4WJ{C8xuqq!v7ytOYMBWlQ8)qY%Jrh1v_^e(Jg?at~FI>QMHDrHTSVFR^1TzVB4;`}iJkH^?t&MxfRRPU2sUsT!m?Nsuv4{0caGPGlAX1Al19kae^X7b8_;Z*1h>u@QLh$2VS-e6fjP9Pe|Jj$j7rbi4xHttOt0v}{a&SHxk9b(WSk)_3T45Z-{bRFG`tGf<j+8vyx^WBsdjaNN~rUqD0306c2goVzp&>)6Z|Ph)cqltP|QsrB3sB0ni$Vt3Rh(BrjF5}#+8Ow#7BY&px8CO(<=2=5cv>M#|pQ8VBSmPa#~mSB#GqmbrE9zytYvnVZ?Q`!Ca>)Xi>uRnjh{5uffK~7d#o&qJW!JRSeFSMObk6HKhk4}MwI$&OP4^NLz$D8Dit*Kt`5G*G$!A_6+r)0Nv&(8Y2<E<4tr29@!`_vLN1dPhf_s-5v$4c6-_kwWMhVjAi$pPDYr^nsn!(+y{e{%LD#{E+emm{nR#^X_!F+Myw>>WIbaj$pE)?)Xte|E@r<1v_mJsn{fpC-aR50v&%|CGCXa(r@lMvpy^%agMw8$LNb>=B}8Fk5ulc0D{f?)IPPYZp3pK%;Yf)&r7}f!#eg>vkFAN92^hUVgg#@%`lUe|)(3gj()>nt&Bo@rtXBf~$4VQB}dQy|}`<4_J2qWs)>s`mtbdu2!+0d``2-hi@%C(1)F$!(5d&Xe^)k+oNVgut<e$^4>;CV49<!#CPzMXA3ZxK>B6^Uyq6pu`fbB6gbU-F9wpC+=PKwv-+?&nwr#m&1S1qi_kuVON;iYDdF~cyOQV;HaJ!a`JeaFP!%bkQ&JGT_#ng=sKSB!;NQN|t?=82|4(}`>iAi_28s0H|I<Fd?h7j$N}l&w5eqd>bsb&tQ#C=Bb23xUdoYon_sPEF_upG+HKBC`hEazU@h@PPKyjG(PN1B61V`$Ko*Ze)`n?R1%MdaG{vPK;#6$a~@XZk6%)~frqMT_sFM6VU+r&lW%H-T7D)hHhAVStE1Xs{raE`UE`q2HW9%I=-gcUQwGjxTp2V!aL7%Va(pMVy|P}rZs<V7F=eFu$FaJn0HR0*x7dIuKB);jR7>H%8&TS~sr`uYw(^~G#UgDm^yv`)bQix@?W#?=&ua7v?ng#&!WeB@dL=mi5(jJ1t!3xdM<0j_oGQ4bgWHtKOrTUFH>WE#p3);HMmEj5`mMty^ljq;sZ`OnPu=tcBS+Rd9O)#?Jzp;y#DY%PT&0wtwT(f_dcLBF;50T+iN#}4~2{cdZ)Ls76}9Umx>4NA1$4n&RcKC?TvEfDRW9I@laE}N|KaaLy=C|xOstf3va+NT(*A)DAZ&69$;qGpb|0zTB2AfXE{8s7z+nSg^f0)l}i#g|3(r6<0?zn=x#C4B8Pxf*){RB({zKhDe`4NfDr#K{Kb04RQn8689rH3htxX>Z=qn+3nwY&k$#K<H}&gjU=69p)Vwincfsje&GI@5m_VWEmxsW#}gul)$)<nw8C&{09*|#ULXV*P^G>u)GVygyN?hn?~S;h^pfJE(~ytgVksCI20{|WLy0wVlgEY_lwdh?^13FEc6`BhhJ7v9tJ83!>*?0v@1KMvKEbCB1LHyB1#h$V79R_BSoYDa3f=?9z|oCtVchOMHtbwentQ)<_^NQoPvKx!DBl|o5sf8qALgR`_9<lG#CdvC>*z9IT{ssrTJiY(B*P6^NUar&9t__8jQoDtU}yD#DOr&pXF&$s1Ub4B84#|8aA6*#2JBaj-9^69_yxU16KsnhARgulbblSW^r0btZ`vPj;vTt(NML>DaRQ60;7kCiTV_Q49Pwe{yt(Dafnk#Qsc7Un^R=7J97h*hWKG%>WTmm+sJZeQII3*ys(Zj)ER&{;X4?qK~o0_QrUsa-5}8X^fNpe9tYdOrYSuYi$XloXjld7Nt%Paw~h6iRTT4B0zb|GL}+Ev?AD?RR#B$7)<<#X5ZrR9v{W3LiUEe$Jp=-prWB!ORyUC~2YpjoVN&8ko7oTp!^Jh(16tC;#hpa5n)E-)1#lu`!nlF4k-z|lS&4(0`^(HKus(>C2u)ug#zh%smLCM%HX~Io0L!I+A1$L_IT)^1peog=<g*mmsro3ZdFTggB2@?j_46tKuV||zBVp1CH(fXq^c57p2~ylljRlzPjILBGD#M4{L95iTALuyk+Tl?SDRK?=-9|Zan%*3$AA|Rc(d}6BKRrve+pd8~42dU(I0p7sF-%DMaG2M_*k4Wq-^vDz-q*?on;~l;5sNf0!@wr!;v4W{91?Z86i#J{(J4Y?6jkd@vSnJ%;8lx6v@xzK@(Yi2U0`EJ5R`#wG~k&-5tG{ZxdNx`JmzRX19Uf7Zk37zTy9-PD@<0)Fvo%9${;qSpO?d)K1NC&hg8*24D_wi3{ngy#cEoFrOmAEaxrl7nsb7Z$z@nt)3rcug9J9`t<j9SG;^&S%l{f>W}8HCmg{kaCa&q^SxOu%i`Q^dcyYKOUh9=WD&v$EoI^$tnl30txN+es*}Q0#XbV=rNxkhxQBo~VSy@^{W%MPKi&V-Wg0`^{?!t~n7EUAuXqcJQXi?F$v=A`5a=oSukGM5px3N+;WJhUgX2$F~BX*_YqAX@QK+$o<rH0sOtzo$cEoYKHf`L9z4}A$cp{J_w&!`VTTySj;7mv(3eSH##k|n(RS>yXvDMis-62?grEq{K*MGIUX5g*X1pF8IC36Lu0%Rr`)ohM*rMN9uClx7zFm|#|gOEb){cVhD7yY!BYC!RM&UPx3HxGL#e1#n^vyqTrYU8Ekerx?*}k(Q~NVNvj?H6G(&HG^sZV-A)U&W2F;6b=0~{mG(xaEwLT1BT7l)=xhnf<OJl;6#H3d<4!y`S7fdVu~e9p@WR=eqo_Bp?+*miXx&UvuZ-XJ1G%Kun`qxZmdvfdfk@EkJf5GTwYlsiRBXe>hhl-vqG^l&Ero>9F<Z+C&i^#$;1+PN5Te&I+{;PVB7+vGqmZwmjV~r)%ZG&$EivsfCs&yf%3}5YB^2(D7NiWT;PI#9D%y`F{CjDTgJP3{r>zPKYrTe0BPaTXh}IaU_pmAuH+x66a|i%I-|h=yJ$$_!(%0hl~JIJp`YJ`rP-)qPiV3g7*<C$&od0y%I;Y$WcRf2;!kyZ7w{P^aPDDr?Z88Uc9i2}qvd<*u=j+**8_rW5WjyerdxT}sUN@?sD(24IT7FN#9!7t>5pL5$*RT6Z?M3(9_D3v*6Y~}Cc47f|FaL{lkB_J{w+V@WZ+YN2BldU6H8Xly`nY^#h-b_YCezdQB_~>QES(G+UOw&Jss;MW~V$9WzMj?IK2bY-gy)?Yp{XEi?Xl{8zCbL%fjYdY7)fMmbQ}z{I#SZ#<^E-*}I9;DG0&Wq9sS?8<xfIfC%r?@;gjkVUYw}-=}Ich)56FBarh8h>bsu!-4f65PDRi1;qh7qB{;)VP_uN+mkZ9FKy7rDGurIdT&+EJEw?<=2M1i${B*m7pNl^UNFzluU27_RWge>tB_vtEW^)0wNN~=ItK6>?QpFFS`bDd;8}Cvm<)do*Q)o>TU58ps3;<7hKQM(K<z)kzmNNMb3n5>pp}vA&KRegTv;Z0=wbQ~W%bx#QjA`A{xQaX{LZiAudy3OWoMi6WLyUpmQ^Yg4UL(fMWr8us}xwTKb>D*V)(!?GmJ(gLt1|rB5wmX$v4kyojN6w_P`b(;L_k{b8jdbmg6_xL_ujc#vludc1LJw>P7}eulu{;D6L?`66S^y6G4w~PVV(Nlc&&$<0osoj?eZ{{A$$gocW#k>&|!MhvP>a#+eyas2OHEQI@R%k+!SM(gGAu5Zz$d1}1*aL(-5(`aI0kkk;CLq(Rnb_+07%3acdgMQfAfPmQdQ#%f^el%xXMz#=a6u8x^0rl_%-!-5sfk-;|}=Pz>S@dfsu+t`hxsYeF0pj0LB&oV357#V?a!F)SHRj2;-9{X&2LfMePB@{IiNPHi2chS*gGrVWo&tUfMhAcaxn%5HM%T>vV8Dy#Tkk6|~$_zApX`)OYreOF4Cf==DO)oqaZr9FCMD-N1Wxk^5+T4$e&_s!SFI4wBLUse2s+`1S%MyWPq}F~JkgD|3t$-Re(cImM8nVtV?q-qClz7P*PRJMkoWOj95(rah8I`e^Ljy4A`ROK1Z*o6dtoi3M45HPNpI3Rz4`e>KXEsn0&I*QZHVM)tNQ1&mPLbeS;>4)U<~m~L3=F-gl!E}!pjd4>9!N(^o0cEYo?}BuymB6&P~y-u;#r%K<y)yW8ftrUm;NwpA<Ztsuj~oYFEo>2(hTFEQnO0kAYRE$vBloNE;JB%j&L#>o&n)$F>P;FYi~1u0ILG{P_m&9rh377H;(KdK=;6bv7v@s+^MY~Zv{UC@<t_t&oRxc*be)@ksF1h1QJ}uDf~E=UQK1|vmbIn)4mF|Lw9a)B?{h=u2*q}528AldpUgcE2WZvlZ#t03X!}XiBKZ?`hE~lA-O?@VaXP5!!T2E_c)wk$^F7wM7XYiqy_5`94>%UEWigsXPMK#r(qe*0LWSfal-`77r-hB@7P9+L4~*FphHh7l4W>$?A17xTO%gXt=^5f23SqIMLir1?g<XRuDe}uZE=km_a<eU0j{mV%|kHvL{teB=OoQJ-cWN$sc|11SD**ElGF)IcIQAzklrQ2$wghCTD0-zCD9D11^#=W##s8WzV(-nn5vp!0!JFJO7>h`_GGD%Jj8U&FQJ_7^Ds-TJWWfnlR&UJ3}$*~xOW(&65Fmgl2zb-oMlvEK!}%}nc<6<ZKGHBmvzci{zjWyE9F9tT3p+bgdTG%G$^_S9()x!QF5!3x4aFa+@>m%^mh0eG@6X<&?pV7ve2&4ojMX;k;M_R&L4r}M%}UF5vS;ynP9ddnB^EM1udepz=H<8(iviL4KOn|k9i;YZv!v)DSJhx-GKapcRhT#-up`t=AG9!xP6S$5`Rd4jpEpUiTe`v9yLU04X=g={G_-W{C05rOAfq<zp3-6bnm99aB*Wy&EuYXv`xj%DMVn`!bw#BIstjy^MjvPxIj-PakZWCJuclT4i?S{Xb9Y+NHSCfYgMQnhXtlwC{J?%$=5}I!ChYa=A2~slE6G<*;k+hIK|Sk&mFQo7c1Pbf-CIyJ4bHLDRX~zivo3>WMK+TTK=qDp&GP;u$bi$S$2%YG{n(P=X1|`Lr9=L&aJ@DZ*lczndTw;Lq&0&6h06GBeLif_`LrvP2oHG3ZC`B)byOV(X*z0A#}&ZJyD=wJ(8`<1#Oi~<Km@3RWDmxosOU(__3${+2Qel`_;kW(fp(j)_13a(<prE4X$1sve54i?qq(@?RVX;{OO^8c;bFF>z;HE&j9RyN|sRY`-7{_fqVYqNN@CD@b1May~uz9tS6@-sMByln@hcn1BD-Q3R|Pv!<sPi925^i%6xE{g9(Z&H%a*xL!!~;t*T@J0`XI8ovxs{gqMkVKBPQa%fIn)5~Not3?ptiXl@nt^P9}ii_q8%dY!aJ&eaFGaaVq=0=}-cdBN2-gt*!^0$1Bv=W20dZD|BCnjn2_v^S+<!uWbUo?%~K2QQp~?Y(etX2jysJT{HcWy4sKUH4w+H!IwB_IaJDV1Ia@flWc{49vg-CMyL5ZbJ*0;Kh-9-Z`ku%-|lEq(E&hy_-B;Wp<CU@_KS1nX1=f>VN?G8nQ~i(%P<Rp=WKqzrHTMwntaz?|`~vhkw23jFBex9|zah9#Q9y^o1jKx$Z8?Nba&V(zt~_FfmNQxDuSU|A25G!LS5F*#f#)Ko{%Xi(c25Bm@OUSw$ajg_{9w2jlu2#?LBBnA#S|>q-!sahhfEnmI*nkraj0{-fD~^QGmUnhh%za&Mlas^N+`=Mi*WT0ZRILX7zELbfC9vn!-;*Qnmps*B(CTIZrxRjE&F9O9{7{GBMSCoW4{qFS2N_hFfvcXS&Oogg*m`)C|B2K)p5%59z?D3Cs!fcrF?Oqfe?O&sioQCBY1_?ZV*J8oO0`hc<2v52_zCCRFBX@mR_!!%2nyowN-_6>cgn5PY@beViI`~eKF(2@=SHXT+f#Ko)=e-)R*(WVU@%dqsn`1z2mY$@mnN}xzOp2F4G7wkwD`X#O~Y{d~(eg%YABEW>(pv)Dd{06lI0OUU4Z(C)zh{~9jXao-_a*9PgF~8|*1O;6cHzU~$Pm_OWHxxQB%hNAWpsjT(5>&`i2Sj*~VQC8{y38v4keGvp4x;V3s6xM?;GzK4s+T^k(c!wdkH|n(uH*yOeqr@!NhDpB1){E>%b@Eb!g~ZY#zbnRM5_0d9+9MHn;0~oAQ?IAYK==^?3iT<-oUDrrMHd(j4>djOlElWvpYlq^O|v6)LB;}+44rNG<)E)%C%$HvS)YJ*E;<O0_P{6I%MkN-fn{uB5RSi|CzZ!m_@^YTuY!_8xX(*xD<xNZpQ}OcX-pGjmaALrt3Jr35c%)fo@@!9D^uc;eQEjNlK^1o4Ly|#{M9SJX$s$%8`ScDf>B1%MvUce8pWRV$neub89j|4{9>u<<!XpRp?}*R#aJ|vPXcC`DvQoPVUnD7N)kl#vv#3RWjqnQhCM+o5=8q4m-hsKJA+)>sf#kr=Ng3f~s=5wth$#n4k99Uo63-o<`Ptzd#e~!{Y6AvM9@}7`%MBSO;LXQAtdM;{#N(HQDwSCgb$G{o@Uwu9K^17AA%Auo2TAE<ao1wM;NUCmtjbV=w5+H!V#x`Hv6P&xKzYhmb_+Zikg*v0nngmRn9QF%8=KktX6bFgJ!2v?ETys><TDBnWGKLYFX$%l0!+#5=H&)pKzRYBHBbvJOt1j3CN=3~?XQ4L1H{h6)uBfW27^4h#%|DE#Liz~L&CRc`n<vP`AhHrVcV+aV>jRAAT0wA|>w>-oTLn^Qo~jr8)3@ERFXn>+_sOOPjNJ^y&|`ty$;DGgzpgaCK`?c&WRK9S^lZ<D-0v8&7Ti}#-{CP<x&ZGf@x$BXmJ4;PcaUVKEVp$f0et`$w`kssG$sd&iDYkbd<0g9N%7#SctLF9X3+ybN9@f5vzki7F0x^HHM#HE+dz;g4K1(?^kwqie26u#b@gI@9G{xXW!984*Ii8^5CMRQ|SMU>RP4y@7j9wEsS+Ep~gnF1BmC*(wW`8NF7|LaP;7@Gk3V4NU}5EjmB`d9xxM6(5LN)OaB7WobX`5Oz7{;C&ybaX$GX8uVLuWpQt9DKp<RI!>;a>IZ6onO11v+)b=L<j$14Y@q+c=D*3PJMX;fo;zmlM$dQ0YR?T`WAhm_zJ84dMqBHYX?$bstW2yTaV|8qIyz`9vI|k;8R&HYX1fX9jZ25UbvS{Q2@6VN7X`}l%yoid*V<g|0smb-xk=UI@~IuN8JcO*A_N4jSC9_z!>(Sep0M3q*BbHXh_433()8YC!3Z8t(NZARy%lmlyr-3Lp5oGq|O!*7!`Rqi!zy%iC-tsH?T?gb4zgj4NR6X=D5UbJ^vDg`8G-K5=%CLrfRk->$`DOb9G{CpJW=kK&M_(zE^%lg=@X2(GML35)#RBX4A&HE^&4%rutIwNUh$s1*3IPX>R=d!_e9aZM*ofc|zS0!%(ed%lsc{#;9SLr*R;pQJVB2Wzy+`L!AK0TBct@&a^`X!3E@6kDxxDm=p%oouWE4qBhtBPh(?^28I^Q!y7u9j6F>K%M3UyLZx4b_BVZPR%QEv&tGZ*jOUzn(8|lpCTLo3BmzBl<USswkcfFLLyiK2|Ek#bCF?~>fe|Ai7Q_sqHPak0XQu|m@uGrGpi_RX&f3q_OcaY1hNAb6D8m#)LsrP8!l^<Py8JMCd-2`X>(3W&X-Pqp9+(anlYs(%OpuyiC2EP~2QK4^vYwXaDKwalQl-fzf_nsG@-H<5ioDRqikLE>aovO-XaPsm#>4z*X^E=K$<ds!7&5N9izo2S(eur2WKL)n%103*fv1wp5Ag(#k9>L+{07fXKN>Ba;>;;JC-ACt^429znFLQGr>|`JiS_RD=MSGmHoUREsj1~1HLd))GoJ~(hL%R@DtCc19%w^OfgEYGp$c~D6FD5HCzj~&$}9lvoJiZiNG)~QNR!ymRMBWlOd^`}&8233Xy>~}Nkl#Hj567>`l#*K+p?@EMt;jOA^OoTd-XNK9o7j~wB4$VmLbf}L)`0+>xN;Gfzl0I_EmyP^=W3{61-g4qqDPmUE1tVe_j}yQnWNktnc7i5A`Z7ZME%8(8d>|Uoj@s(^kZUK>qN?2>H~EuIU;xI|zm5iZ^MLO@^NsT0A@eCYDuE=>mO#K_kml0K0{U+9gRWZa1N|t)_3z4>u@t=kxL^oqss69{A&9n>|>aUaIXDqkS92!tWm)8|HTq-7u$$n1(0H=(9+RK~BsDGAdV}VYI+xtj!ct&|J52N3qR$Ap=oVvYFx*47H)QYHTSc1f@4fAk8j|0^nBjliqDalhzViJdPc9yKLO&qRtXpfSoP0IcsK04FqfblQwu&r!)szZkEBp9o7()n?Y~DN~d`a{9<tSFeFph@Vbb84TnHHJRVS;Vsj6Z4dTOlbgH*SSWXBlO01%%N7s6wQr=w@l#55}UjAN$=$Af<{Zf?$$YAclVGm=V1FmwUd4QWz$>wP<zLv$;PZqz)qkyw=hbqwP(!NE)aM<;Z<oTg^5^O~wm~jE&N$jsd+tp@s_^fT~%>BEGustWV4-%JOzJ&2wj9ipNC5((LGvw2*XgBm~Iguo%?Mz2q$DFI*0&wz=R6at^ak3StGfP0kbChL%5>ny&G{r=RW*DT**SaKO<m-mY;`#`bWv@#H!Ejp{_&Bu+mmes7zzd=x1LDZSfAYWhyddUU;*&`FyT=qL5DF|tDhS1zR#O_<>f<WG=|yPp#LQvUNGm!{q^C42Fe&BsmO26S8P?%0P~&QyVT+)E@s4L7;|ex0@hr{Ow)sullBSR7lbh1Os0n>sy^d1v4bC-}Kh_~r;GWC5sP>kA2|WC#4;SAf3;(k6<EUH@&$=#cFv6Yny7`1t*92y7ipt@S0rlY}+L^C7?Cf6e!R*V=-t_p8VgjK?cy1h*St;g#0iPh`GJ6zl&^$r1k3>4Bwo{ekH?ABpV=1L380g=Q#;v=oC2+ar^r68V0U9@GDKapzG#t^|=Un=Rod-r|(Cr5}!_~8{uJ=f>w1N^}2--*PjDXe-X`0?sY>-aArC<@CeoM#E;_h97F%kvG;|73Rtd6w}G;R@v)0XydF{Bl7{@{{AG3q^5UcA3Xqu=zvRUl!Yc*$*i{TnJ#LbTHQ%~J0|8I;r~1EZW^dIe7kK@+0*Qo)o(Tc8TgV!AL3nXWghEVNSh_xG<8zgQ<TwFpH!tWg8=co)F1pdKA2lJqj~{HUZP*$<#vK#AG5u#y~W_$?Pjm%K&K;#77y&%tr!zBrQlJCw!jiIQBW2T(iwZh(ircG;+>?tytxl-1s?(f(J#$w9Y&g2~Qb9Uh$=O%Lg@=bypMLyu>thoOH!kMsG-ybEvJ;)nZRb-Vtte@q4a<D=vLDc3nU?M>$lKfwR!@%VVw_mB3+6;8Mq1{ekDj1mL98Li!2@Ilx1eh_`xr*6=qSy8T$8b$FL)9i=VK8D~s9IPKu;N7j_@vnd!9b%e6uY1(Z?q6k?GM(HEdVTnTO9VG`K>2`-fmc&nm>Kl4dvNsPD6qZ?j_^P6p@YX_t%|{EcE4YtkI4&L(9aloEahSF_8jnq4~RN&(~Uf}q*NQt%ao=vaLI+_(&_rV6OT3(fa*!_%nBS8D#<vQEmB+vx%KuH(O6NQFQPcG5fLHlf+hR}X6r3I-~!wYer7?WTrOlsLtLvuEQ0k?LR70_jS&?V!+@-CV+gzd`ig-L{8gE@4A;rY$*b8aFTk82vM&56e^sWd*`mW~Faz9)(M}jf$I)daDH$b!DX7qMerNCZgME{VxE7tp#V(92%Wm%aJPz+)-S`=@+AI9gxy#`f{sRQbZ=wXK0p0Fud<bluZfSkGD$6us1h8nG8UKRSW6WWJIY<LL!-dry;ZoT;;Bx}t$FT;KFM`++Di?0x%R*5|`>)v7$~FkAiTYQ)!{gcE!7EaQrr>vLf_|uJj1wR?X$?|1cg%*<G!FKSAr&?WQrKmBxB@Aso$EwJgPCA@DpxzP5|?cDcNrxeAu4_B1$XwV(`nD|`R!e3pvJH@fxr}|ofWkXi8hGh7gcqd25X}T)-h-xO>vw}$$VyuI1Wo?=?}#=Ch?jkb*!<9WBCr0LCy@Lzm+zoR84ZRVCiSJu$O4UmZoJC45TuS0#_X*5yt8@{Ssep)!SRQ+E%Mx(mq#_s+Qj4f^i}PO7tCG@<BvR<!ZeYFDrA)ZK>UB)Mlk?R)1$&NLu~=Kh^rdGg{~BWD7jXY%gVS(fK8I_$5p#YlwMcih^7iiunz`0;V{RBPfsgBQ-Z&k(nXl#R>4|?^Z`Z^Wtz`;;9(M6!%hoHBC<g#(z{W$~4nJEM&jM*F~exwI1govbL@FmeRFJk5HewbghGqqHDJ1m^Bd|OOHqo*%g2k32;;JAU7!%ywZGMnX~$!@Y0`T1oJEJn-~fMbVN|Be-%H0_Q_4T7;9BzntFO0enRCYtJaHP@aAvA8aZgMPlRKs!t6Da5B}R9)5ujH4&;kob*_qpHIL$WD<)W=wJAJyy??kB;66=5`;8(g&vmGyt=<vLKxE<Ww}C><Z-1igr8ZL+W9hEwALar|jPS#WQ{{rlg+i%T|Lxy|DuoB3KOJMRTgC?bhF))f+u-yY0K)PLzQ6U0g*`fPt)nK*wecg2h^+jCgcZn*=%$XUfPCGnUbl>=(Y*Zp*d~=^cFRNyAu^Ofy^nzQYr08X(6eQpBrUk_$ZfFn)}uo_hXmLBAg{m?m(=K=c-EJ;_?<Mr>ji`bgQIjLJQt-EMagz$G{?h`t}t&|^$}-sFG`XyfA{%^D+Ia!TlEDdknUhi5b*SFv#pFl-B<Werwo^v`xs(E{c4#MgFcwgxGe#981&F!c6;;OQEv|Djdz6aj*qmW=qkE>d`*4^8scF7z4GJeCh35nE{nm8j?|RK_=&Q`tNw|NyPLa++@0K1z2pitJGUXxwbFtt&(yAHUtxrYS-UcAtbKH{_8qe*88`LPZ1J>LxUCnyl?CB1BVJXx#bc^;1pzN;J)+;O5fm3{Yk>LCkX9r(=VV78gIO3WG@Nu05luBm0cwud^1?2A9~W9WL=5QMPa+ujFAr+K9<9r_KdCmT;Ikp1d;gP0u;<plW$jUc;FDFJiv?7wKmZ}qgj5;s#$f@kjAS&e1EISf&CYMECE1LN)#~1Eu7}+Ao|wwM^+a^_(dg;Ms)D7cP_0414QT>K-t4xjsxJVu%ENsmvyes9HP){U6{^2xGf%#G)b*TjAZ4f=g;Izb2obY8qgL@0>SWWGx@(!<CAzE)oVq&3Bmq2{Vpp9k%m1ICn8}~Tl@^Zc31|1{=t$`24%p=sFi~RsYP65K=<+T8`yRxLPL!h`Kw|v6Kd$C7!Bpn(3pI-8wQbSynXhRWi;CAYj5e_wbk#=`t5xh`j(nEWJ*T#;*kZ~E13yj3r3!OQ>_LtezE#yPc5BdWgTHTiQ<&t_%In5hv$oEIOUIh4CX6BFcvBhiQbhd1@&hAs?q^_zgu<N?*j)z72dU)bv8^9M!8iMX{Qd`lup?}pI$lkjpKNcH{f@1*L@^{&-FM;b513>j_D1qEo)g%xa=T;4(d(|CsM|9vWV7ylPu0a^rVS@if!qOwHZWpArVgha9i-?zUKawC_;zR=qbst;-)sKe&|b>eGPN8|TxN)(@Uj(H{F)qNaOWqnB=|26!ro$dn9CP@LOkt0sD$pm^Z1i%9FdP9plY%dOMrTr+W|OU)lZHD?duCE90a<Dky^w5?|bmKH*U(hi1gL0ll`i9jj86*lu;P~vs!{;Exb=L;tjK(B4wIYQI<ygO>v^|@OGHmit#QWMZ9+d%&jydsogj!wpZm041KewFeJ2bsHFc<7mA1iRv69v6srE<BCsvZRbrxaE-9fQL#V#|&k>$0IpS>?o$=TI@)FQ1%F29q9DROj?ZHnNHm96a+zBMMR^XZ5eD(xi)xB6}i%{EjclmYz%&nSv{HdM<*Dz!d;RqvMe(^l?=+Xc|3QRA>qaEZ;;^l8a1dsstS3|;}scw0CAo)vs7bL&OtP>p%JQ)eo+;=0)AJ|y+qT-TIhf`jXWBu2ERaWgcR^RyjEqFvbV0X^zf4;aaUD*Dd?9!)OdAl^SdIz4w++byq8B{6TO>y(~$a!n*Tu@&XF+<}PPaAxhbbFk<3QDU$vmU(Ga~;}Y^=hzcQEMZX{WP@x?}*vKb1;*@svMH$dPK{LgrL848&v8WsaLS*F`FtdY`^wpIguS<VZKtJVnduM`lnAnzW4Y9ylB3*nT?!EnR7l!nuzq1+Vt@BWWe?Lh((QDyt+WQP3}wOFMJfc%2c!YkFsR&8WGHvs;7NWnv}KyOQg$e-gxCyTb2qirva$QrL(aOr#|~>ThzFMQ@bC5-->e&X$JtmG2(g?Xo9w}$b2g4{<%hSj*)_+Wt?uQE-H#D10(%d6xyC~5AdTTY@0=^Q!+!oP^6kD(R4_N0rh;cDvzwCWqLEAP{f8*FiwJMo)xKjQ#;<-oPPPd4$(;aISl6MjCdZ5#2AySRpmeA=2>CUT*Dss`T#k0fFG<d5P`|Bt*w}CAWUs$oM<znWiI6~#V9e&#Ib%PClqllTI+SK5^xC}bto`=qBC+2oB2cHFklUih_r9I1Vrt4+;Xg|kz9w3?=n6bsRl#>V;?GADMTs7@=({pAc$$e1NpQiFd`SYGT&HT?9f_0fsgYoOBB7wUt-n%H=r9wp?R!FVP=P$*!1u?k^8P_7oOMM-_j`FWkDyKk~n?=s%Uqh(Cj0Qd}>WxY{*<6F}cNm`p-t^R&83;Z5{J~B~~L(WR5DJza=BIk|(8EHl~BtoW7PcPQ1n#kAA1(Y#6|py_8z`*ntM=?Hb=oAN%-;69fgIUMrxU0cb-i>g{h$Ic=fIqq$Ho#;BT%IHWe=O`x{iu}(WaExE->Y^dTKdo@?-+i?dQGQ72#^?kiY{%bK?giC*d98Oc!z%tJ;Fw;`yfjR@rz}MI=irvD4f(Dj2jKTPP7fy?aL5KqwjL$KjEize%O}GQgJb%F`EH?xXtR~0Hv|epvU1Mx{w~k8Ll{<WN(lzR&Ip#)$!Q^gxR&cHK;;HOlK|BocPH2-4-Y-IX*)INm`JKgp#F6PK?~cV&U_O5T#^N0|;!1Fh|2Aa_|0W~K00y>k&WVSu#OGcVna_iBy?1_jarOF5W1qg*tTLe>BA0x7eAH6Rk8gkg<zgw$tuRkT@x*Z`&3a(u2I+Ew4kcmIZHo5<6ZG+0zq`5F$ubQ@SAg{Eprt9fF+%7`jM`&`!)|xukc7#XC{L56*tOUoBcEYO!X}Qs@%WXL))o~^C)Hq;B=aV+++@-ymqo}!Sn$K;)z*?=_uoe4gb%nuffB@Tuk&c8o@+-+7zZiv4^u(%JN?89qGSrqpi8siAO`lfjQeXF`^26+(LmdAFZL#ydyTgqL7;6xW=KdFa0JcDhAww@R@%SEHn!?w696$=dZj<@b@75I&@m*W7)}>=JE)#2eKB_Qn{ed%oyvnC;4R08wE%ME#UicSUmAvHZS#nBgxfuxxJPFz+3H9o+hzz3f}jw)D`V?VpF511M(o^z!yDX$yqQiM`d4kFJJ3x#K!b9o@A2O1ULV`W8?aAZYrZc-=ALv^M#_{LKf2cKuzG!quiADX$PInd*q@PMiOYdB>b<reO@U0f6?zT>Dvs(KSlV_#Jk0Q>`Xc&<N-kUYq#V7o$JULH`wwa`l>sn*w5mx!r7HiAY<S$XZxKvwRt5v(?l2DN8M|;rgr;8XxranZWH2TKe+5_=CpiN8ZcH2=m}~Gfup!=oLDs;Q9hW;O5qW|q9p^0f@3k2*a+9Js*mlE_D$*LNj)@eI!NHYvpSUMaTpo!Bh7_x%6{I2WyRL3sqWGCq^?SSPLF9XkowZ_Kx5?_p{q{=IG#giZBqDCuCUl-w;CJbG5W!qtP{1olpX#_8Xko<&%RwfzQq<wr#b8lM%ijWnS)OLqUQPI6Q%r(F5l&S^zeP}j7;UgpHXV-oWWA@aC1lYRxBrRj2ob&Q7ZQpSW5DJ@wr1;2+p(EQk`+VUUt?2ls9`OQJH3OtNvM}KO+~$ivI6zub<<E*aJ+d`>IV%o1*(PQ6&?ak^q7J+z@IjCl$Q&N37t30%XUMxS3kWg>X)NZ{*t;N_10-zan4XBb8%i@G2t9KZ8AmJ8{-DFwOD!Rd`>ebjkwlfZLcTZ+So827a>IiHXKpxPn}k$EGN@hZ<OgKkOxN|O%;P|^4>?ptIY?;UgFt8wsy$|yc}3|2mXvyvtXor%`d3$<Jwu&2uJha(<LbUa8ezv)NZ9#XD9q$Yll`{+!Cv9SbPJM&!FlCZCA7Pv=_IV^;4UM*FDp5?XX+5O*_3a?N!9B)h3n6Yt7RX>#;R5k_5{5X8bsq=^L@z3_MAxDmEu8Zz()cvATv~S*uk)`9?beOkQfN66$1(N)^#g>41srNe;*3xbX-CHjf7XeP`Jje7LjxHJnlH_@cIx)9~0TjTUGwt6y5~A@_MuKhW(70XO)SL4XyaH#g=jKx!z|*+si})2bp^f+60(t^m2#rf$<=Tu-yYHqg8I;l#o6Bs^V>C{T0{q=IW67p~rq>e4WlSGVa+G|rW(K1m7K0&(-)E6=Li(3U=<jv{&QEj<{0BSRD3wsdAVq(;V-1CQyY;!Xi_(?N1s{jx5@HqD4ZSkz+%q8_^6J>0x0A_$a#+I3YUFpuIHF}vQ?Q#(%|fb~5YOTfkm-kI|xyyBNt7y>@L!dwyCzvtsvmD<MO0(NgSaILPIDR>lICd9yS(%BV_mUh)ir%_-n3iQlRVD{qDc#7xJwYdJBuGa;}pU&4|Bp3cr+~tO9<O}{{)2rsd*QO@M%QeNon4s#@(xxDYgw*x=c<?(o^+)*caeub~t$bHqG{}q_>!FeAKzXNu7m<eXuwFphhbM$W@6<IUm}WY4TfqRYs_OJM&GieDlA8#7x-&N2BNH3Z#owiM3JR<;4xe1_{kemOT^E+FF6p2)+J?>td(Wf1C`}$Y{QoN6!Dhya50y5jw;~j`{1wkKqzzNyL1<cHx`EYD%i`o#UfOG&lwFtuoX1*3we1ujg{~>XRWDji&1!Dt)DBsCq)pUD6jYxNv5uGkTj$PA>&%t9sh^OigEliPsye!nGSVxI_1y%tRcrggG+yRmo?6uK3UTS%u*2Gxv4z_roX#hu8~V=~!#kGNmx9&!Ja;u%Bm_MFo^U*fmogwJ3qalsF7MbeRys!1Ghbf#TU+s!Rph1F#+nqkfLxIZjOwSjCw@E;w?pHxPnt}bMh*cUn`=RukH*17iJOLz1gQq574Gk_xHRGG5x`otMI^jB1X;`eLengZX&sZUu~B6Vqn}DF<+hOAu8NH`0At5IHFrx&mb|N7UevBNv3o{Vg%xic!lelHu6~Q7Ad>p->Rl&)NQQEB7@Tt$#&ia1)oURW^nq$i=GU%5yA9U^aq}0IlXu<nEm!6}xmPS%=ltg#p#Er{f?QqWBL-96;Nz;(t?>ddy7RvMA~Ed<Y?GO^4~CP2g93)5uvecoFA339ad<(lN@Pt{A>Tg{g7&|NO8}LA27M9D!a`j;MLmh=f*C_;P+lJNrFH<R5<1cTs1nq1{WhS6)DGo4T_@}FLIqS;Ib&^36_+T=D-?!LDu)(4C8#yes!v$~Z6LHe`zl^Am4QChmT1_OhByE;ep{cvMOiU(6Oubfw{fdzW8!9*b@BEK;x|({3BKd6J0*@{!ClhV1QG-52DcRWc+zVw>#OY-6(VL5AMZRQYK$0=Mm2Q{T7wsASA5Y`uRNM~4_xU-xR#))eRaQ8rT2~a3`}$8%xQ-E{0Ex7H?T3LNcRTagJ$3l3nGLHyZU#@D+9qF%``j091Dm(H1byFCC|`IiAGk881R7@5#36ulR$<(nS##K0<2<-RJ}H?bsfQoR`<#MF2cTxonf^09CJdy*;PcrOg-hAu;rYw-5S{l)m6;ALwIdt<fq?^_?)@=ZE+N&=^w##+4oMJTJ%L&iQ>2fEdqmg#~GKzxj0$B5!{LD<TW#*vn3*<*}%VJRHi3_GGdGQmZ(hqLOkB>E-q!({WpQ+j$s%9r^~^%-m`-*Y{onz-Xae)pl%}t`mW6JXp3IZHA@>p7ys`E8z@6TX5Q|eShkyaCu0MKaaZXgGKZzfr*23cY1s#cTWCGy6w)2s)wYxJE|n;Vxsp0dw{xr8aH<_C|MQEDxTlNvhwIlN1A??CM89rM>-aKCF#iB=Lh%#gxY|{6w$xGxNMj+<+Q_)oZiI!#<F@HtG*qoWtCkO&OV!x0RuXTtt1UmP!>;K%JF`LQm0j;8A&5g!R-kB6y`atLhvdW1>huKf6M+7nHVXEr;A11EjMC3VD37XstDYj7K1v=P1XSB>&A0#5BMDkMcuR(uyK~T?SlwVSGz2{wfn2HpgMO``l!zwOX&fbFlvP&&D)IStXtmyQr~?r!f)K=GOOp9S5>Fh)OwVR~qfJF>HicAROv<`FfbW{u*9AFDtfqp(8;kL%YGqjb?{__g3L@vz&^*|Cx@1Ro$f6FZR%)u!G}&~^PMs_<FY9_k1qtX@F|s-`%uDu#N0a4gvPx!==pVfj7YFk%e+q<by+RJg9&a_}Vl&2^gb{O{=$2}8c~amaN2q=k2Lc#D8M+%>l%d~R??$p8dLbEN6di04HOB;@=%$X&sx^_SWoBt+bNq5=8(TsWqKc5(gtY}NA+2M+Luf~N)#7n-GLl|%3$?f0!zfGhmMX9q;+46njNKN+p%Qg91?pp`y>56AQrZbX?aCG6Bqi%&-rQ!4JWIyq;zeD?H!o@kX~K-W=O<NE0eQ}K9Bv5<OB85<dvh_uM35BklC1;gSklvwTgvI6WlEd`PYdJ{kb{8m1=ZNJBa?1ujbqHEhd*>vt4)cXDu1jx1*UrAmuWo2m1E}#uK+P8a0-fYzXX~t_eiDalb!I-bV>HepBAboGBwWXXq}NSY2^Ytv%N3R@5<A2l@?g0w9WPaUp)rds^Z%0@yhf8S7<`_snnPJROF)tmL3%$cx^#KRcK?3ka|WS`8(}5)~m^A(72KAqTwy0GV}chy{^bqQ=S5Oy=K<-em#o}JjN)|PxgxD20rLmj`^N+wSZ7}nlUU;>wP^yj3)qWsHTPci8?Y$sD=E@;3ZeA(M<%&{u#0{+#1ywfKE08Ni*t2vn48BlXr}THcdw}Y(6i5p&}!jEX^QAZV_#>$AaSxdjOv6o{Bl_q4jRjNfZd(V=mRwtM=<iCrZI%74>67VMAq88~Te4m2{m*%eUI^+7?NsuDD3Os!@@>+6mE8Q!q_cL%cA(4V>;H-uYagJL!-jHQ~{v7e<FSYHUO8<BK+o=|jb#CUT(c_8$Kiuk@ks"
MODULE_SOURCES = json.loads(zlib.decompress(
    base64.b85decode(PACKED_MODULE_SOURCES)).decode("utf-8"))
for module_name in ("colorlib", "adjust_colors", "notebook_workflow"):
    module = types.ModuleType(module_name)
    module.__file__ = f"{module_name}.py"
    sys.modules[module_name] = module
    exec(compile(MODULE_SOURCES[module_name], module.__file__, "exec"),
         module.__dict__)

from notebook_workflow import *

PALETTE_SLUG = palette_slug(PALETTE_SLUG)
if IN_COLAB:
    WORK_DIR = Path("/content/rang-workflow") / PALETTE_SLUG
else:
    WORK_DIR = Path.cwd() / "cache" / "notebook-upload" / PALETTE_SLUG
WORK_DIR.mkdir(parents=True, exist_ok=True)
RECIPE_PATH = WORK_DIR / f"{PALETTE_SLUG}-recipe.json"
use_arial()

def receive_file(local_path, label, canonical_stem=None):
    if IN_COLAB:
        print(f"Choose {label} from your computer")
        uploaded = files.upload()
        if len(uploaded) != 1:
            raise ValueError("Upload exactly one file")
        filename, data = next(iter(uploaded.items()))
        uploaded_name = Path(filename).name
        if canonical_stem:
            suffix = Path(uploaded_name).suffix.lower()
            if not suffix:
                raise ValueError("The uploaded file needs a filename extension")
            output = WORK_DIR / f"{canonical_stem}{suffix}"
        else:
            output = WORK_DIR / uploaded_name
        output.write_bytes(data)
        return output, uploaded_name
    if not local_path:
        raise ValueError(f"Enter a local path for {label}")
    path = Path(local_path).expanduser().resolve()
    if not path.exists():
        raise FileNotFoundError(path)
    if canonical_stem:
        if not path.suffix:
            raise ValueError("The source file needs a filename extension")
        output = WORK_DIR / f"{canonical_stem}{path.suffix.lower()}"
        if path != output.resolve():
            output.write_bytes(path.read_bytes())
        return output, path.name
    return path, path.name

def load_workflow_zip(local_path):
    archive, _ = receive_file(
        local_path, "the workflow ZIP from the previous notebook"
    )
    with zipfile.ZipFile(archive) as handle:
        total = 0
        for item in handle.infolist():
            member = Path(item.filename)
            if member.is_absolute() or ".." in member.parts:
                raise ValueError("The workflow ZIP contains an unsafe path")
            total += item.file_size
        if total > 200 * 1024 * 1024:
            raise ValueError("The workflow ZIP expands beyond 200 MB")
        handle.extractall(WORK_DIR)
    if not RECIPE_PATH.exists():
        raise FileNotFoundError("The uploaded ZIP does not contain the expected recipe")
    return archive

def make_workflow_zip():
    archive = WORK_DIR / f"{PALETTE_SLUG}-workflow.zip"
    return make_workflow_archive(RECIPE_PATH, WORK_DIR, archive)

def offer_download(path):
    print("Saved:", path)
    if IN_COLAB:
        files.download(str(path))

print("Working folder:", WORK_DIR)
```
