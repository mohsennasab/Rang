# Rang examples in Python

This notebook shows how to use Rang for categorical and continuous data with pandas and matplotlib. It installs Rang from GitHub and downloads a public dataset, so it can run in a fresh Google Colab session.

```python
import matplotlib.pyplot as plt
import pandas as pd
from matplotlib.colors import ListedColormap

import rang

plt.rcParams.update({"figure.dpi": 120, "axes.spines.top": False, "axes.spines.right": False})
rang.list_palettes()
```

## 2. Load public online data

We will use the [Palmer Penguins dataset](https://allisonhorst.github.io/palmerpenguins/). It contains measurements for three penguin species observed in the Palmer Archipelago, Antarctica.

```python
data_url = "https://raw.githubusercontent.com/allisonhorst/palmerpenguins/main/inst/extdata/penguins.csv"
penguins = pd.read_csv(data_url)
print(f"Loaded {len(penguins)} rows from the online dataset")
penguins.head()
```

## 3. View the palette ramps

Calling `rang.rang(name)` returns the complete stored ramp. The order shown here is the order used for continuous interpolation.

```python
names = rang.list_palettes()
fig, axes = plt.subplots(len(names), 1, figsize=(10, 1.25 * len(names)))

for ax, name in zip(axes, names):
    colors = rang.rang(name)
    ax.imshow([range(len(colors))], cmap=ListedColormap(colors), aspect="auto")
    ax.set_title(name, loc="left", fontsize=11)
    ax.set_axis_off()

fig.suptitle("Rang palette ramps", fontsize=15, y=1.03)
fig.tight_layout()
plt.show()
```

## 4. Use a discrete palette for categories

For a small number of categories, `rang.rang()` follows the stored pick order and returns well separated colors. Here Golestan distinguishes the three penguin species.

```python
species_mass = (
    penguins.dropna(subset=["species", "body_mass_g"])
    .groupby("species", as_index=False)["body_mass_g"]
    .mean()
)
colors = rang.rang("Golestan", len(species_mass))

fig, ax = plt.subplots(figsize=(8, 4.5))
ax.bar(species_mass["species"], species_mass["body_mass_g"], color=colors, edgecolor="white")
ax.set(title="Average penguin body mass", xlabel="Species", ylabel="Body mass, g")
ax.grid(axis="y", color="#dddddd", linewidth=0.7)
ax.set_axisbelow(True)
plt.show()
```

## 5. Use a continuous color map for measurements

`rang.cmap()` creates a native matplotlib color map. Termeh is a sequential ramp designed for continuous measurements such as elevation, depth, and the body-mass values used below.

```python
plot_data = penguins.dropna(subset=["bill_length_mm", "flipper_length_mm", "body_mass_g"])

fig, ax = plt.subplots(figsize=(8, 5.5))
points = ax.scatter(
    plot_data["bill_length_mm"],
    plot_data["flipper_length_mm"],
    c=plot_data["body_mass_g"],
    cmap=rang.cmap("Termeh"),
    s=42,
    alpha=0.9,
    edgecolors="white",
    linewidths=0.35,
)
fig.colorbar(points, ax=ax, label="Body mass, g")
ax.set(
    title="Penguin measurements with the Termeh color map",
    xlabel="Bill length, mm",
    ylabel="Flipper length, mm",
)
ax.grid(color="#e5e5e5", linewidth=0.6)
ax.set_axisbelow(True)
plt.show()
```

## 6. Inspect the artwork source

Every palette carries provenance metadata. This makes it possible to keep the visual result connected to the artwork that inspired it.

```python
rang.source("Termeh")
```
